```python
import pandas as pd
import numpy as np


LANGUAGES = ["en", "ru"]
files_to_load = [
    "../data/generation step/multilingual_gpt54mini_answers.csv",
    "../data/generation step/multilingual_gpt54mini_answers_advanced.csv",
    "../data/generation step/multilingual_gemma4_answers_advanced.csv",
    "../data/generation step/multilingual_gemma4_answers.csv",
]
MODELS_TO_USE = ["gpt54mini", "gpt54mini", "gemma4", "gemma4"]
PROMPTS_TO_USE = ["simple", "advanced", "advanced", "simple"]

all_dfs = []
for file, model, prompt in zip(files_to_load, MODELS_TO_USE, PROMPTS_TO_USE):
    df = pd.read_csv(file)
    df = df[df["language"].isin(LANGUAGES)]
    df["model"] = model
    df["prompt"] = prompt
    all_dfs.append(df)

all_samples = pd.concat(all_dfs, ignore_index=True)
```

```python
columns_to_keep = [
    "model",
    "prompt",
    "language",
    'query_id',
    'distractor_id',
    'difficulty',
    'query',
    'gold_judge_text',
    'answer_with_distractor'
]
all_samples[columns_to_keep].to_csv("../data/manual_labeling_samples.csv", index=False)
```
